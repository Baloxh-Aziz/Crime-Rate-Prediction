## 2. Data Cleaning
    - Dropping irrelevant columns and handling missing values.

```python

df = df.drop(columns=['X Coordinate','Y Coordinate', 'Location',
        'G', 'Case Number', 'Block', 'IUCR', 'Description',
        'Beat', 'Ward', 'Community Area', 'FBI Code', 'Year', 'Updated On'
    ])
```

## 3. Feature Engineering
    - Extracting Month, Hour, DayOfWeek from the Date column.

```python
df['Date']=pd.to_datetime(df['Date'],format='mixed')
df['Year']=df['Date'].dt.year
df['Month']=df['Date'].dt.month
df['Hour']=df['Date'].dt.hour
```

```python
df['DayOfWeek'] = df['Date'].dt.dayofweek
df['Domestic'] = df['Domestic'].astype(int)
df[['DayOfWeek','Domestic']].head()
```

## 5. Model Building
### Label Encoding
    - Converting categorical columns to numbers.

```python
from sklearn.preprocessing import LabelEncoder
le_crime = LabelEncoder()
le_location = LabelEncoder()
```

```python
df['CrimeTypeCode'] = le_crime.fit_transform(df['Primary Type'])
df['LocationCode'] = le_location.fit_transform(df['Location Description'])

X = df[['Month','Hour', 'DayOfWeek', 'District', 'Domestic', 'CrimeTypeCode', 'LocationCode']]
y = df['Arrest']

print('Feature Shape', X.shape)
print('Arrest rate:' , y.mean().round(3))
```

### Train/Test Split
    - 80% training data, 20% testing data.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=42)
```

### Baseline Model: Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X_train, y_train)
```

### Main Model: Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    min_samples_leaf=10,
    class_weight='balanced',
    random_state=42
)
rf.fit(X_train,y_train)
```

```python
y_pred = rf.predict(X_test)
```

## 6. Model Evaluation
    Accuracy: 0.84 | ROC-AUC: 0.86

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import roc_auc_score
y_proba = rf.predict_proba(X_test)[:,1]
print('Accuracy :', round(accuracy_score(y_test, y_pred), 3))
print('ROC-AUC  :', round(roc_auc_score(y_test, y_proba), 3))
print('5 Probablities: ', y_proba[:5].round(2))
```

```python
from sklearn.metrics import confusion_matrix
import numpy as np
cm=confusion_matrix(y_test,y_pred)
color_matrix = np.array([[1, 0],   # TN=green, FP=red
                          [0, 1]]) # FN=red,   TP=green

sns.heatmap(color_matrix, annot=cm, fmt='d',
            cmap=sns.diverging_palette(10, 133, as_cmap=True),
            vmin=0, vmax=1,
            xticklabels=['0','1'], yticklabels=['0','1'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
from sklearn.metrics import classification_report

y_pred = rf.predict(X_test)
y_proba = rf.predict_proba(X_test)[:,1]

print('Accuracy: ', round(accuracy_score(y_test,y_pred), 3))
print('Roc-Auc: ', round(roc_auc_score(y_test,y_proba), 3))
print(classification_report(y_test,y_pred,target_names=['No Arrest','Arrested']))
```

### Feature Importance

```python
from sklearn.metrics import precision_score, recall_score, classification_report

# Sabse pehle test data par predictions nikalain
y_pred = rf.predict(X_test)

# Precision aur Recall calculate karein
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print('Accuracy: ', round(accuracy_score(y_test,y_pred), 3))
print('Roc-Auc: ', round(roc_auc_score(y_test,y_proba), 3))

print(f"Precision Score: {precision:.4f}")
print(f"Recall Score: {recall:.4f}")

# Ek complete report ke liye ye behtareen hai:
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
